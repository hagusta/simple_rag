```python
import arxiv
import os

download_dir = "/home/hagusta/.arxiv_downloads"
os.makedirs(download_dir, exist_ok=True)

#paper_id='2511.07587'
paper_id='2512.20605'

print(f"Attempting to download paper with ID: {paper_id}")
print(f"Saving to directory: {os.path.abspath(download_dir)}")

try:
    # Construct a search object specifically for a list of IDs
    search = arxiv.Search(id_list=[paper_id])

    # The client results method returns a generator; you can iterate over it
    # to get the first (and in this case, only) result
    first_result = next(arxiv.Client().results(search))

    # The Result object has helper methods for downloading
    # download_pdf saves the file to the specified directory with a standardized filename
    # You can also use result.download_source() to get the source files (.tar.gz)
    path = first_result.download_pdf(dirpath=download_dir)

    print(f"\nSuccessfully downloaded: {first_result.title}")
    print(f"File saved to: {path}")

except StopIteration:
    print(f"Error: Could not find paper with ID {paper_id}")
except Exception as e:
    print(f"An error occurred during download: {e}")
```

```python
import arxiv
import os

def search_and_download_arxiv(query_string, max_results=10, download_dir='arxiv_papers'):
    """
    Searches arXiv for papers matching a query and downloads their PDFs.

    Args:
        query_string (str): The search query (e.g., "machine learning applications").
        max_results (int): Maximum number of papers to retrieve.
        download_dir (str): Directory where PDFs will be saved.
    """
    print(f"Searching arXiv for query: '{query_string}' with {max_results} results max...")

    # Construct the API client and search object
    client = arxiv.Client()
    search = arxiv.Search(
        query=query_string,
        max_results=max_results,
        sort_by=arxiv.SortCriterion.SubmittedDate,
        sort_order=arxiv.SortOrder.Descending
    )

    # Perform the search and get results (generator)
    results = client.results(search)
    
    # Ensure the download directory exists
    if not os.path.exists(download_dir):
        os.makedirs(download_dir)
        print(f"Created directory: {download_dir}")

    download_count = 0
    for result in results:
        try:
            # The download_pdf method handles the file saving
            # It names the file automatically using title and ID
            path = result.download_pdf(dirpath=download_dir)
            print(f"Downloaded: '{result.title}' to {path}")
            download_count += 1
        except Exception as e:
            print(f"Error downloading '{result.title}': {e}")
            continue

    print(f"\nSearch complete. Successfully downloaded {download_count} PDFs to the '{download_dir}' directory.")


```

```python
search_and_download_arxiv(
        query_string="all:RAG AND cat:cs.LG", 
        max_results=200, 
        download_dir=download_dir
    )
```
